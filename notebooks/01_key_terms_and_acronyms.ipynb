# Key Terms and Acronyms

This notebook is used as a simple reference list for the main engineering, uncertainty, simulation and verification terms used throughout the analysis.

# Glossary

| Term / acronym | Meaning in this project |
|---|---|
| 95% interval assumption | The assumption that the stated uncertainty range represents an approximate 95% interval around the nominal load. This is used to parameterise the truncated normal uncertainty model. |
| Base load | The nominal heat load assigned to a consumer before scenario-specific factors are applied. |
| Capacity | The available cooling duty for a subsystem. |
| Capacity margin | The difference between available capacity and calculated load. |
| Cascade | The system arrangement where load passes downstream from HVAC to CW, then from CW to FW, and finally to the Ultimate Heat Sink. |
| Cascaded load | Load propagated from an upstream subsystem into a downstream subsystem. |
| CI | Confidence interval. A range used to express sampling uncertainty in an estimated probability or statistic. |
| Consumer | An individual item of equipment that places a heat load on one of the cooling subsystems. |
| Convergence | The process of checking whether Monte Carlo estimates become stable as the number of simulations increases. |
| CW | Chilled Water system. This subsystem receives its own local consumers plus the total propagated HVAC load. |
| Dependency model | A simulation assumption where some consumer loads are allowed to move together, representing shared demand drivers. |
| Deterministic baseline | The fixed-load calculation used as the reference case before uncertainty is applied. |
| DoE | Design of Experiments. In this project it is used to explore influential consumers and interactions, not to estimate failure probabilities directly. |
| EDA | Exploratory Data Analysis. The initial checks used to understand the structure, completeness and behaviour of the input data. |
| Fractional factorial design | A reduced DoE structure that studies selected combinations of high and low input settings without testing every possible combination. |
| FW | Fresh Water system. This subsystem receives its own local consumers plus the total propagated CW load. |
| Gaussian copula dependency modelling | A method used to introduce dependency between sampled uncertain loads while preserving each individual load distribution. |
| Global Gaussian copula | A dependency model where a common correlation structure is applied across the sampled loads. |
| HVAC | Heating, Ventilation and Air Conditioning system. This is the first subsystem in the cooling cascade. |
| Independent model | A simulation assumption where consumer load uncertainties are sampled independently from each other. |
| Interaction effect | The combined influence of two inputs where their joint effect differs from simply adding their separate effects. |
| ITEM_000955 | The workbook item used to pass the HVAC total load into CW for the assessed scenarios. |
| ITEM_001103 | The workbook item used to pass the CW total load into FW for the assessed scenarios. |
| Lf | Load factor. A scenario-specific multiplier used to represent whether a consumer is on, off or partially loaded. |
| Load-to-capacity ratio | The subsystem load divided by the available capacity. Values closer to 1 indicate less margin. |
| Local load | The load generated by consumers directly connected to a subsystem, before adding any cascaded upstream load. |
| Logistic regression | A classification model used where enough overload events exist to estimate which inputs increase the chance of overload. |
| Main effect | The individual influence of changing one input or consumer load on the model output. |
| MC | Monte Carlo simulation. A repeated random sampling method used to estimate the distribution of possible cooling loads. |
| Non-degenerate row | A load row with a meaningful uncertainty range. Rows with no variation cannot contribute to correlation or sensitivity ranking. |
| Overload | A condition where the simulated load exceeds the available subsystem capacity. |
| Overload probability | The estimated probability that a subsystem exceeds its available cooling capacity in the Monte Carlo simulation. |
| P95 | The 95th percentile. In this project it is used to describe high-end simulated loads or spare capacity behaviour. |
| P99 | The 99th percentile. This is used to examine more extreme but still plausible simulated outcomes. |
| Physical bounds | The minimum and maximum sampled values allowed for a load so that the simulation remains physically meaningful. |
| Propagation row | A workbook row used to represent transferred subsystem load rather than a separate physical consumer. |
| Random seed | A fixed value used to make random simulation results reproducible. |
| rho | The correlation parameter used in the Gaussian copula dependency modelling approach. |
| Robustness | Assessment of whether the conclusions remain stable under reasonable changes in modelling assumptions. |
| Rule of three | A simple upper-bound check used when no overload events are observed. With 100,000 simulations, the 95% upper bound on overload probability is approximately 0.003%. |
| Scenario Alpha | One assessed operating state with its own set of active consumers and scenario-specific factors. |
| Scenario Bravo | The second assessed operating state with a different set of active consumers and scenario-specific factors. |
| Scenario load | The load used in a scenario after applying the relevant load and utilisation factors. |
| Sensitivity analysis | Analysis used to identify which consumers or assumptions have the strongest influence on simulated subsystem risk. |
| Simulation iteration | One complete random draw of all uncertain loads and the resulting propagated subsystem totals. |
| Spare capacity | The remaining unused capacity after the subsystem load has been applied. |
| Spearman correlation | A rank-based correlation measure used to assess monotonic relationships between sampled inputs and risk-related outputs. |
| Standardised coefficient | A coefficient scaled so that different inputs can be compared on a more consistent basis. |
| Subsystem | One of the main cooling systems being assessed: HVAC, CW or FW. |
| Subsystem-block Gaussian copula | A dependency model where correlation is applied within subsystem groups rather than across all loads equally. |
| Triangular distribution | An uncertainty model defined by minimum, most likely and maximum values. |
| Truncated normal distribution | A normal distribution restricted within lower and upper physical bounds. In this project, the spread is based on the 95% interval assumption. |
| Uf | Utilisation factor. A further multiplier applied to the consumer load in the assessed scenario. |
| Ultimate Heat Sink | The final external reservoir that receives the heat rejected from the FW system. |
| Verification | Checks used to confirm that the model has been implemented correctly. |
