# Key Terms and Acronyms

This notebook is used as a simple reference list for the main engineering, uncertainty, simulation and verification terms used throughout the analysis.

# Glossary

<table align="left">
  <thead>
    <tr>
      <th align="left">Term / acronym</th>
      <th align="left">Meaning in this project</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td>95% interval assumption</td>
      <td>The assumption that the stated uncertainty range represents an approximate 95% interval around the nominal load. This is used to parameterise the truncated normal uncertainty model.</td>
    </tr>
    <tr>
      <td>Base load</td>
      <td>The nominal heat load assigned to a consumer before scenario-specific factors are applied.</td>
    </tr>
    <tr>
      <td>Capacity</td>
      <td>The available cooling duty for a subsystem.</td>
    </tr>
    <tr>
      <td>Capacity margin</td>
      <td>The difference between available capacity and calculated load.</td>
    </tr>
    <tr>
      <td>Cascade</td>
      <td>The system arrangement where load passes downstream from HVAC to CW, then from CW to FW, and finally to the Ultimate Heat Sink.</td>
    </tr>
    <tr>
      <td>Cascaded load</td>
      <td>Load propagated from an upstream subsystem into a downstream subsystem.</td>
    </tr>
    <tr>
      <td>CI</td>
      <td>Confidence interval. A range used to express sampling uncertainty in an estimated probability or statistic.</td>
    </tr>
    <tr>
      <td>Consumer</td>
      <td>An individual item of equipment that places a heat load on one of the cooling subsystems.</td>
    </tr>
    <tr>
      <td>Convergence</td>
      <td>The process of checking whether Monte Carlo estimates become stable as the number of simulations increases.</td>
    </tr>
    <tr>
      <td>CW</td>
      <td>Chilled Water system. This subsystem receives its own local consumers plus the total propagated HVAC load.</td>
    </tr>
    <tr>
      <td>Dependency model</td>
      <td>A simulation assumption where some consumer loads are allowed to move together, representing shared demand drivers.</td>
    </tr>
    <tr>
      <td>Deterministic baseline</td>
      <td>The fixed-load calculation used as the reference case before uncertainty is applied.</td>
    </tr>
    <tr>
      <td>DoE</td>
      <td>Design of Experiments. In this project it is used to explore influential consumers and interactions, not to estimate failure probabilities directly.</td>
    </tr>
    <tr>
      <td>EDA</td>
      <td>Exploratory Data Analysis. The initial checks used to understand the structure, completeness and behaviour of the input data.</td>
    </tr>
    <tr>
      <td>Fractional factorial design</td>
      <td>A reduced DoE structure that studies selected combinations of high and low input settings without testing every possible combination.</td>
    </tr>
    <tr>
      <td>FW</td>
      <td>Fresh Water system. This subsystem receives its own local consumers plus the total propagated CW load.</td>
    </tr>
    <tr>
      <td>Gaussian copula dependency modelling</td>
      <td>A method used to introduce dependency between sampled uncertain loads while preserving each individual load distribution.</td>
    </tr>
    <tr>
      <td>Global Gaussian copula</td>
      <td>A dependency model where a common correlation structure is applied across the sampled loads.</td>
    </tr>
    <tr>
      <td>HVAC</td>
      <td>Heating, Ventilation and Air Conditioning system. This is the first subsystem in the cooling cascade.</td>
    </tr>
    <tr>
      <td>Independent model</td>
      <td>A simulation assumption where consumer load uncertainties are sampled independently from each other.</td>
    </tr>
    <tr>
      <td>Interaction effect</td>
      <td>The combined influence of two inputs where their joint effect differs from simply adding their separate effects.</td>
    </tr>
    <tr>
      <td>ITEM_000955</td>
      <td>The workbook item used to pass the HVAC total load into CW for the assessed scenarios.</td>
    </tr>
    <tr>
      <td>ITEM_001103</td>
      <td>The workbook item used to pass the CW total load into FW for the assessed scenarios.</td>
    </tr>
    <tr>
      <td>Lf</td>
      <td>Load factor. A scenario-specific multiplier used to represent whether a consumer is on, off or partially loaded.</td>
    </tr>
    <tr>
      <td>Load-to-capacity ratio</td>
      <td>The subsystem load divided by the available capacity. Values closer to 1 indicate less margin.</td>
    </tr>
    <tr>
      <td>Local load</td>
      <td>The load generated by consumers directly connected to a subsystem, before adding any cascaded upstream load.</td>
    </tr>
    <tr>
      <td>Logistic regression</td>
      <td>A classification model used where enough overload events exist to estimate which inputs increase the chance of overload.</td>
    </tr>
    <tr>
      <td>Main effect</td>
      <td>The individual influence of changing one input or consumer load on the model output.</td>
    </tr>
    <tr>
      <td>MC</td>
      <td>Monte Carlo simulation. A repeated random sampling method used to estimate the distribution of possible cooling loads.</td>
    </tr>
    <tr>
      <td>Non-degenerate row</td>
      <td>A load row with a meaningful uncertainty range. Rows with no variation cannot contribute to correlation or sensitivity ranking.</td>
    </tr>
    <tr>
      <td>Overload</td>
      <td>A condition where the simulated load exceeds the available subsystem capacity.</td>
    </tr>
    <tr>
      <td>Overload probability</td>
      <td>The estimated probability that a subsystem exceeds its available cooling capacity in the Monte Carlo simulation.</td>
    </tr>
    <tr>
      <td>P95</td>
      <td>The 95th percentile. In this project it is used to describe high-end simulated loads or spare capacity behaviour.</td>
    </tr>
    <tr>
      <td>P99</td>
      <td>The 99th percentile. This is used to examine more extreme but still plausible simulated outcomes.</td>
    </tr>
    <tr>
      <td>Physical bounds</td>
      <td>The minimum and maximum sampled values allowed for a load so that the simulation remains physically meaningful.</td>
    </tr>
    <tr>
      <td>Propagation row</td>
      <td>A workbook row used to represent transferred subsystem load rather than a separate physical consumer.</td>
    </tr>
    <tr>
      <td>Random seed</td>
      <td>A fixed value used to make random simulation results reproducible.</td>
    </tr>
    <tr>
      <td>rho</td>
      <td>The correlation parameter used in the Gaussian copula dependency modelling approach.</td>
    </tr>
    <tr>
      <td>Robustness</td>
      <td>Assessment of whether the conclusions remain stable under reasonable changes in modelling assumptions.</td>
    </tr>
    <tr>
      <td>Rule of three</td>
      <td>A simple upper-bound check used when no overload events are observed. With 100,000 simulations, the 95% upper bound on overload probability is approximately 0.003%.</td>
    </tr>
    <tr>
      <td>Scenario Alpha</td>
      <td>One assessed operating state with its own set of active consumers and scenario-specific factors.</td>
    </tr>
    <tr>
      <td>Scenario Bravo</td>
      <td>The second assessed operating state with a different set of active consumers and scenario-specific factors.</td>
    </tr>
    <tr>
      <td>Scenario load</td>
      <td>The load used in a scenario after applying the relevant load and utilisation factors.</td>
    </tr>
    <tr>
      <td>Sensitivity analysis</td>
      <td>Analysis used to identify which consumers or assumptions have the strongest influence on simulated subsystem risk.</td>
    </tr>
    <tr>
      <td>Simulation iteration</td>
      <td>One complete random draw of all uncertain loads and the resulting propagated subsystem totals.</td>
    </tr>
    <tr>
      <td>Spare capacity</td>
      <td>The remaining unused capacity after the subsystem load has been applied.</td>
    </tr>
    <tr>
      <td>Spearman correlation</td>
      <td>A rank-based correlation measure used to assess monotonic relationships between sampled inputs and risk-related outputs.</td>
    </tr>
    <tr>
      <td>Standardised coefficient</td>
      <td>A coefficient scaled so that different inputs can be compared on a more consistent basis.</td>
    </tr>
    <tr>
      <td>Subsystem</td>
      <td>One of the main cooling systems being assessed: HVAC, CW or FW.</td>
    </tr>
    <tr>
      <td>Subsystem-block Gaussian copula</td>
      <td>A dependency model where correlation is applied within subsystem groups rather than across all loads equally.</td>
    </tr>
    <tr>
      <td>Triangular distribution</td>
      <td>An uncertainty model defined by minimum, most likely and maximum values.</td>
    </tr>
    <tr>
      <td>Truncated normal distribution</td>
      <td>A normal distribution restricted within lower and upper physical bounds. In this project, the spread is based on the 95% interval assumption.</td>
    </tr>
    <tr>
      <td>Uf</td>
      <td>Utilisation factor. A further multiplier applied to the consumer load in the assessed scenario.</td>
    </tr>
    <tr>
      <td>Ultimate Heat Sink</td>
      <td>The final external reservoir that receives the heat rejected from the FW system.</td>
    </tr>
    <tr>
      <td>Verification</td>
      <td>Checks used to confirm that the model has been implemented correctly.</td>
    </tr>
  </tbody>
</table>
