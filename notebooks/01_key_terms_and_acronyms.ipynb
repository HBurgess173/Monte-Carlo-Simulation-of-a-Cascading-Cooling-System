# Key Terms and Acronyms

This notebook provides a short reference list for the engineering, data, and modelling terms used throughout the project.

The project overview is covered in the README. This notebook is therefore intended to support the later analysis notebooks by defining the terminology before the calculations begin.

## Cooling System Acronyms

| Term | Meaning | Use in this project |
|---|---|---|
| HVAC | Heating, Ventilation and Air Conditioning | First subsystem in the cascade. HVAC loads are passed downstream into CW. |
| CW | Chilled Water | Second subsystem in the cascade. CW receives the HVAC load and adds its own local loads. |
| FW | Fresh Water | Final cooling subsystem in the cascade. FW receives the total CW load and adds its own local loads. |
| Ultimate Heat Sink | External heat rejection destination | Represents the final external reservoir that receives the total FW heat load. |
| Subsystem | One part of the cooling system | The project assesses HVAC, CW and FW separately. |
| Consumer | Equipment item placing heat load on a subsystem | Each consumer contributes a load that may be active or inactive in each scenario. |

## Load and Scenario Terms

| Term | Meaning | Use in this project |
|---|---|---|
| Base load | Nominal heat load for a consumer | The starting load value before scenario factors and uncertainty are applied. |
| Lf | Load factor | Scenario-specific factor that represents whether a consumer is on, off, or partially loaded. |
| Uf | Uncertainty factor | Workbook factor applied to the base load in the original scenario load calculation. |
| Scenario Alpha | First assessed operating state | One set of active consumers and load factors. |
| Scenario Bravo | Second assessed operating state | Alternative operating state with a different subset of active consumers and load factors. |
| Local load | Load added by consumers directly attached to a subsystem | Used to avoid double-counting propagated upstream loads. |
| Cascaded load | Total load after upstream loads have been propagated downstream | Used for the final subsystem-level capacity assessment. |
| Propagation row | Workbook row representing an upstream load already passed downstream | These rows are excluded from local downstream sums and replaced by recomputed cascade totals. |
| ITEM_000955 | HVAC-to-CW propagation row | Represents the HVAC total embedded in the CW workbook load. |
| ITEM_001103 | CW-to-FW propagation row | Represents the CW total embedded in the FW workbook load. |

## Margin and Capacity Terms

| Term | Meaning | Use in this project |
|---|---|---|
| Capacity | Available cooling duty for a subsystem | Compared against deterministic and sampled subsystem loads. |
| Capacity margin | Difference between capacity and load | Positive margin means the subsystem remains below capacity. |
| Spare capacity | Remaining capacity after load is applied | Used to show how close each subsystem gets to its capacity limit. |
| Load-to-capacity ratio | Load divided by capacity | A ratio below 1 is below capacity; a ratio above 1 indicates overload. |
| Overload | Load exceeds available capacity | Main subsystem-level risk event assessed by the Monte Carlo simulation. |
| Overload probability | Proportion of simulations where load exceeds capacity | Used to express cooling adequacy as a probability rather than pass/fail only. |

## Uncertainty and Monte Carlo Terms

| Term | Meaning | Use in this project |
|---|---|---|
| EDA | Exploratory Data Analysis | Initial review of workbook structure, missing values, load counts and load distributions. |
| Deterministic baseline | Fixed-value engineering calculation | Reference case where nominal loads are compared directly with capacity. |
| MC | Monte Carlo | Simulation method that repeatedly samples uncertain loads to estimate risk. |
| Simulation iteration | One sampled realisation of the system load state | Each iteration produces sampled HVAC, CW and FW loads. |
| Random seed | Fixed number used to initialise random sampling | Allows simulation results to be reproducible. |
| Triangular distribution | Distribution defined by minimum, most likely and maximum values | Used as one uncertainty model for sampled consumer loads. |
| Truncated normal distribution | Normal distribution restricted to physical lower and upper bounds | Used as the second uncertainty model. |
| 95% interval assumption | Assumption used to infer normal spread from uncertainty bounds | Treats the lower and upper uncertainty bounds as an approximate 95% range. |
| Physical bounds | Minimum and maximum allowed sampled values | Sampling checks confirm generated values remain inside these limits. |
| P95 | 95th percentile | Load value exceeded by approximately 5% of simulations. |
| P99 | 99th percentile | High-end load value used to assess near-worst-case capacity demand. |
| CI | Confidence interval | Range describing uncertainty in an estimated probability. |
| Rule of three | Approximate upper bound when no overloads are observed | Used to interpret zero observed overload events. |

## Dependency Modelling Terms

| Term | Meaning | Use in this project |
|---|---|---|
| Independent model | Sampling assumption where each consumer varies separately | Used in Notebook 05 as the first Monte Carlo model. |
| Dependency model | Sampling assumption where loads can rise or fall together | Used in Notebook 06 to test whether dependency changes risk. |
| Gaussian copula | Method for introducing correlation while preserving each load distribution | Used to model dependent demand between uncertain loads. |
| $\rho$ | Latent Gaussian copula correlation | Controls the strength of dependency between sampled loads. |
| Global Gaussian copula | Dependency model with one common factor across all loads | Represents system-wide common demand behaviour. |
| Subsystem-block Gaussian copula | Dependency model with common factors within HVAC, CW and FW blocks | Allows loads to be correlated within each subsystem. |
| Non-degenerate row | Load item with a non-zero uncertainty range | Only these rows vary during sampling. |

## Sensitivity, Verification and Results Terms

| Term | Meaning | Use in this project |
|---|---|---|
| Sensitivity analysis | Assessment of which inputs most influence the result | Used to identify important consumers and assumptions. |
| Spearman correlation | Rank-based correlation measure | Used when overload events are rare or absent. |
| Logistic regression | Model for estimating relationships with a binary event | Used where there are enough overload events to analyse drivers of overload. |
| Standardised coefficient | Regression coefficient after input scaling | Allows consumer effects to be compared on a common basis. |
| DoE | Design of Experiments | Used as a screening approach to identify influential assumptions and interactions. |
| Fractional factorial design | Reduced experimental design covering selected factor combinations | Used to explore main effects and selected interactions efficiently. |
| Main effect | Effect of changing one factor | Used to rank individual modelling assumptions. |
| Interaction effect | Combined effect of changing two factors together | Used to identify assumptions that matter jointly. |
| Verification | Checking that the model has been implemented correctly | Includes cascade conservation, bounds checks and monotonicity checks. |
| Robustness | Checking whether conclusions are stable under modelling choices | Includes uncertainty model comparison, dependency comparison and convergence. |
| Convergence | Stability of Monte Carlo estimates as sample size increases | Used to check that 100,000 simulations is adequate for interpretation. |
